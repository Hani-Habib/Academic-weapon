```python
print("1.1")
%pip install git+https://github.com/alemartinello/dstapi 
import pandas as pd 
import matplotlib.pyplot as plt 
from dstapi import DstApi

print("loading data from Statistikbanken")
IFOR41 = DstApi("IFOR41") 
IFOR32 = DstApi("IFOR32")

IFOR41.tablesummary(language="da")
IFOR32.tablesummary(language="da")

print(IFOR41.variable_levels("ULLIG", language="da"))
print(IFOR41.variable_levels("KOMMUNEDK", language="da"))
print(IFOR32.variable_levels("DECILGEN", language="da"))
print(IFOR32.variable_levels("KOMMUNEDK", language="da"))

print("Retrieving the Gini-coeficient for Danmark") 
gini_params = {
    "table": "IFOR41",
    "format": "BULK",
    "lang": "da",
    "variables":[
        {"code": "ULLIG", "values": ["70"]}, 
        {"code": "KOMMUNEDK", "values": ["000"]}, 
        {"code": "Tid", "values": ["*"]}
    ]
}

gini = IFOR41.get_data(params = gini_params)
print(gini.head())
print(gini.info())

gini["TID"] = pd.to_numeric(gini["TID"])
gini["INDHOLD"] = pd.to_numeric(gini["INDHOLD"])

gini_table = gini.rename(columns = {"TID": "ÅR", "INDHOLD": "Gini"})
gini_table = gini_table[["ÅR", "Gini"]].sort_values("ÅR", ascending = True)
print(gini_table)


print("plotting the Gini-coeficienten for Danmark")
plt.plot(gini_table["ÅR"], gini_table["Gini"])
plt.xlabel("År")
plt.ylabel("Gini")
plt.title("Udvikling i Gini-koeficient for Danmark")
plt.show()


print("retrieving the deciles for Danmark")
table_params = {
    "table": "IFOR32",
    "format": "BULK",
    "lang": "da",
    "variables":[
        {"code": "DECILGEN", "values": ["*"]},
        {"code": "KOMMUNEDK", "values": ["000"]},
        {"code": "Tid", "values": ["*"]}
    ]
}

deciles = IFOR32.get_data(params = table_params)
print(deciles.head())
print(deciles.info())

deciles["TID"] = pd.to_numeric(deciles["TID"])
deciles["INDHOLD"] = pd.to_numeric(deciles["INDHOLD"])


print("creating a pivot table with deciles as columns and years as rows")
deciles_table = deciles.pivot(index ="TID", columns = "DECILGEN", values = "INDHOLD").reset_index()

deciles_table = deciles_table.rename(columns = {"TID": "ÅR"})
print(deciles_table)
print(deciles_table.info())

print("calculating the top 10% income share")
deciles_columns = [column for column in deciles_table.columns if column != "ÅR"]
deciles_table["Top 10% andel"] = (
     deciles_table["10. decil"] / deciles_table[deciles_columns].sum(axis = 1) * 100) 

plt.plot(deciles_table["ÅR"], deciles_table["Top 10% andel"])
plt.xlabel("År")
plt.ylabel("(%)")
plt.title("Udvikling i top 10% andel for Danmark indkomst")
plt.show()


print("calculating the correlation between Gini-coefficient and top 10% income share")
print("by merging the the two tables and thne calculating the correlation between the two")
merged_table = pd.merge(gini_table, deciles_table, on = "ÅR")
correlation = merged_table["Gini"].corr(merged_table["Top 10% andel"])

print("printing out the correlation")
print(correlation)

print("The two measures are highly correlated and both show the same trend in income inequality over time ind Denmark.")


print("1.2") 

import numpy as np
from scipy.optimize import minimize

print("creating and defining the variables")
gini_table["ÅR"] = pd.to_numeric(gini_table["ÅR"])
gini_table["Gini"] = pd.to_numeric(gini_table["Gini"])

t = gini_table["ÅR"].to_numpy() - 1987
y = gini_table["Gini"].to_numpy()

print("defining the linear and quadratic models") 

def linear_prediction(beta, t):
    return beta[0] + beta[1] * t

def quadratic_prediction(beta, t):
    return beta[0] + beta[1] * t + beta[2] * t**2
print("calculating the error from the two models")
def linear_error (beta, t, y):
    return np.sum((y - linear_prediction(beta, t))**2)

def quadratic_error(beta, t, y):
    return np.sum((y - quadratic_prediction(beta, t))**2)
print("now we calculate the betas wich minimize the error for the two models using scipy.otimize")
linear_initial_beta = [0, 0]
quadratic_initial_beta = [0, 0, 0]
linear_result = minimize(linear_error, linear_initial_beta, args=(t, y))
quadratic_result = minimize(quadratic_error, quadratic_initial_beta, args=(t, y))

print("we print the the beta values for wich the error is minimized.")

print("Linear model coefficients:", linear_result.x)
print("Quadratic model coefficients:", quadratic_result.x)

print("we solve the same minimize problem with np.polfit and compare the results.")

linear_coefficients = np.polyfit(t, y, 1)
quadratic_coefficients = np.polyfit(t, y, 2)
print("Linear model coefficients (np.polyfit):", linear_coefficients)
print("Quadratic model coefficients (np.polyfit):", quadratic_coefficients)

print("we now seek to predict the Gini coefficient for 2030 and 2040 with the two models.")

t_future = np.array([2030, 2040]) - 1987
linear_predictions = linear_prediction(linear_result.x, t_future)
quadratic_predictions = quadratic_prediction(quadratic_result.x, t_future)
print("Linear model predictions for 2030 and 2040:", linear_predictions)
print("Quadratic model predictions for 2030 and 2040:", quadratic_predictions)

print("we plot the results of the predictions")

plt.scatter(t_future, linear_predictions, color="blue", label="Linear Predictions")
plt.scatter(t_future, quadratic_predictions, color="red", label="Quadratic predictions")
plt.xlabel("Year")
plt.xticks(t_future, [2030, 2040])
plt.ylabel("Gini Coefficient")
plt.title("Predictions for Gini Coefficient in 2030 and 2040")
plt.legend()
plt.show()

print("Both specifications capture historical upwards trend in the gini coefficinet, however the quadratic model is a more flexible model and therefore perhaps more suitable for predicting future values. however both model predictions should be trusted with caution, as they are based on historical data and do not account for possible economic- and policy changes in the future.") 

print("1.3 Municipalities")



print("Retrieving the Gini coeficients for all municipalities")
gini_mun_params = {
    "table": "IFOR41",
    "format": "BULK",
    "lang": "da",
    "variables": [
        {"code": "ULLIG", "values": ["70"]},
        {"code": "KOMMUNEDK", "values": ["*"]},
        {"code": "Tid", "values": ["*"]}
    ]
}

gini_mun = IFOR41.get_data(params=gini_mun_params)


print("Retriving tth decile data for all municipalities")
decile_mun_params = {
    "table": "IFOR32",
    "format": "BULK",
    "lang": "da",
    "variables": [
        {"code": "DECILGEN", "values": ["*"]},
        {"code": "KOMMUNEDK", "values": ["*"]},
        {"code": "Tid", "values": ["*"]}
    ]
}

decile_mun = IFOR32.get_data(params=decile_mun_params)


print("Remove all of Denmark")
gini_mun = gini_mun[gini_mun["KOMMUNEDK"] != "Hele landet"].copy()
decile_mun = decile_mun[decile_mun["KOMMUNEDK"] != "Hele landet"].copy()

gini_mun["INDHOLD"] = pd.to_numeric(gini_mun["INDHOLD"])
decile_mun["INDHOLD"] = pd.to_numeric(decile_mun["INDHOLD"])

print("Make decile table")
decile_mun = decile_mun.pivot(
    index=["KOMMUNEDK", "TID"],
    columns="DECILGEN",
    values="INDHOLD"
).reset_index()

deciles = [f"{i}. decil" for i in range(1, 11)]

decile_mun["Top 10 share"] = (
    decile_mun["10. decil"] /
    decile_mun[deciles].sum(axis=1) * 100
)

print("Merge Gini and top 10 share")
mun_data = pd.merge(
    gini_mun[["KOMMUNEDK", "TID", "INDHOLD"]],
    decile_mun[["KOMMUNEDK", "TID", "Top 10 share"]],
    on=["KOMMUNEDK", "TID"],
    validate="1:1"
)

mun_data.columns = ["Municipality", "Year", "Gini", "Top 10 share"]
mun_data["Year"] = pd.to_numeric(mun_data["Year"])


print("1. Correlation across municipalities")

latest_year = mun_data["Year"].max()
latest = mun_data[mun_data["Year"] == latest_year]

correlation_mun = latest["Gini"].corr(latest["Top 10 share"])

print("Correlation across municipalities:", correlation_mun)
print("Correlation over time:", correlation)


print("2. Most and least unequal municipalities")

most = latest.nlargest(10, "Gini")
least = latest.nsmallest(10, "Gini")

print("\n10 most unequal:")
print(most[["Municipality", "Gini"]])

print("\n10 least unequal:")
print(least[["Municipality", "Gini"]])

ranking = pd.concat([most, least]).sort_values("Gini")

plt.barh(ranking["Municipality"], ranking["Gini"])
plt.title("Most and least unequal municipalities")
plt.xlabel("Gini coefficient")
plt.show()



print("3. Change in Gini")

first_year = mun_data["Year"].min()

first = mun_data[mun_data["Year"] == first_year]
last = mun_data[mun_data["Year"] == latest_year]

change = pd.merge(
    first[["Municipality", "Gini"]],
    last[["Municipality", "Gini"]],
    on="Municipality",
    suffixes=("_first", "_last")
)

change["Change"] = change["Gini_last"] - change["Gini_first"]

largest = change.nlargest(10, "Change")
smallest = change.nsmallest(10, "Change")

print("\nLargest changes:")
print(largest[["Municipality", "Change"]])

print("\nSmallest changes:")
print(smallest[["Municipality", "Change"]])

change_plot = pd.concat([largest, smallest]).sort_values("Change")

plt.barh(change_plot["Municipality"], change_plot["Change"])
plt.title("Largest and smallest changes in Gini")
plt.xlabel("Change in Gini coefficient")
plt.show()


print("The correlation across municipalities is 0.975, which is very close to the correlation over time of 0.982. Both correlations are very high and positive, which means that municipalities with a high Gini coefficient also tend to have a high share of income going to the richest 10 percent. The two measures therefore give a very similar picture of inequality.") 

print("The figure shows that Gentofte, Rudersdal and Vejen are among the most unequal municipalities, followed by Hørsholm, Lyngby-Taarbæk, Frederiksberg and Copenhagen. At the other end, municipalities such as Faaborg-Midtfyn, Egedal, Bornholm and Faxe have much lower Gini coefficients.")
print("A general pattern is that several municipalities around Copenhagen and some of the larger cities have relatively high inequality. However, Vejen stands out, since its Gini coefficient is much higher than most other municipalities. This suggests that there could be extreme incomes in small populations, that can have a large effect on the Gini coefficient in a municipality.")
print("The largest increase in the Gini coefficient is found in Vejen, followed by Rudersdal, Gentofte and Lyngby-Taarbæk. Copenhagen, Hørsholm, Fredensborg, Aarhus and Frederiksberg have also experienced relatively large increases. Vejen again stands out with a particularly large change, while the change in Fanø is very small.")



print("1.4 Extension")

api_edu = DstApi("LIGEUB1")

params_edu = {
    "table": "LIGEUB1",
    "format": "BULK",
    "lang": "da",
    "variables": [
        {"code": "BOPOMR", "values": ["*"]},
        {"code": "HERKOMST", "values": ["*"]},
        {"code": "HFUDD", "values": ["*"]},
        {"code": "ALDER", "values": ["*"]},
        {"code": "KØN", "values": ["*"]},
        {"code": "Tid", "values": ["2024"]}
    ]
}

education = api_edu.get_data(params=params_edu)

education["INDHOLD"] = pd.to_numeric(
    education["INDHOLD"],
    errors="coerce"
)

print("Keep totals for origin, age and sex")
education = education[
    (education["HERKOMST"] == "I alt") &
    (education["ALDER"] == "Alder i alt") &
    (education["KØN"] == "I alt")
]


print("Total population")
total = education[
    education["HFUDD"] == "I alt"
][["BOPOMR", "INDHOLD"]]

total.columns = ["Municipality", "Total"]


print("Long education + PhD")
high = education[
    education["HFUDD"].isin([
        "H70 Lange videregående uddannelser, LVU",
        "H80 Ph.d. og forskeruddannelser"
    ])
]

high = high.groupby("BOPOMR")["INDHOLD"].sum().reset_index()
high.columns = ["Municipality", "High education"]


print("Share with high education")
edu = pd.merge(total, high, on="Municipality")

edu["High education share"] = (
    edu["High education"] / edu["Total"] * 100
)


print("Merge with Gini")
extension = pd.merge(
    latest[["Municipality", "Gini"]],
    edu[["Municipality", "High education share"]],
    on="Municipality"
)


print("Correlation")
corr_edu = extension["Gini"].corr(
    extension["High education share"]
)

print("Correlation:", corr_edu)


print("Figure")
x = extension["High education share"]
y = extension["Gini"]

a, b = np.polyfit(x, y, 1)

plt.scatter(x, y)
plt.plot(x.sort_values(), a * x.sort_values() + b)

plt.xlabel("Highly educated (%)")
plt.ylabel("Gini coefficient")
plt.title("Education and income inequality")

plt.show()

print("We examine the relationship between the share of highly educated people and income inequality across municipalities. The correlation is 0.674, which shows a positive relation. Municipalities with a larger share of highly educated people therefore tend to have higher Gini coefficients. The figure also shows this positive trend, although there are some outliers. This may reflect that highly educated people often have higher incomes, which can increase income differences within a municipality.")



import matplotlib.pyplot as plt

print("EXERCISE 2")
print("")

print("#### 2.1 ####")

# -- given parameter values --

N = 50_000  # number of individuals

age_start = 18   # Starting age
age_retire = 65  # retirement age

# Education: short, medium, long
educations = ['short', 'medium', 'long']
p_e = np.array([0.40, 0.35, 0.25])
S_e = np.array([1, 3, 5])
h_e0 = np.array([1.00, 1.20, 1.55])
delta_e = np.array([0.010, 0.020, 0.030])

# human capital parameters
delta = 0.06
sigma_psi = 0.10

# job-finding parameters
lam = 0.60   # job-finding probability
sigma = 0.05  # job-separation probability

# income parameters
y_SU = 0.45    # under education
rho = 0.6     # replacement rate while unemployed
y_floor = 0.35  # floor for those never employed

# Random number generator
rng = np.random.default_rng(2026)

n_years = age_retire - age_start  # 47 years, ages 18-64

# Human capital shocks
print("All global variables and shocks are defined")

print("")
psi = rng.lognormal(-0.5 * sigma_psi**2, sigma_psi, size=(N, n_years))
print(f"mean value of psi is validated: {round(psi[1,:].mean(), 2):.4f}")

print("")
print("#### 2.2 ####")
print("")
print("# 2.2.1")

# Draw each individual's education and corresponding years of schooling
education_level_i = rng.choice(3, size=N, p=p_e)  # 0=short, 1=medium, 2=long
S_e_i = S_e[education_level_i]                     # each individual's own years of education

# Check education shares against p_e
short_count = 0
medium_count = 0
long_count = 0

for i in range(N):
    if education_level_i[i] == 0:
        short_count = short_count + 1
    elif education_level_i[i] == 1:
        medium_count = medium_count + 1
    else:
        long_count = long_count + 1

print("Comparing education shares to target p_e:")
print(f"Share short:  {short_count / N:.4f} (target {p_e[0]})")
print(f"Share medium: {medium_count / N:.4f} (target {p_e[1]})")
print(f"Share long:   {long_count / N:.4f} (target {p_e[2]})")


uniforms = rng.uniform(0.0, 1.0, size=(N, n_years))

def simulate_employment(S_e_i, uniforms):
    employed = np.empty((N, n_years))
    for i in range(N):
        for t in range(n_years):
            if t < S_e_i[i]:
                # still in education: not employed
                employed[i, t] = 0
            elif t == S_e_i[i]:
                # just finished education: enters labor market as unemployed
                employed[i, t] = 0
            else:
                # already in the labor market
                if employed[i, t - 1] == 0:
                    # was unemployed -> finds a job with probability lambda
                    if uniforms[i, t] < lam:
                        employed[i, t] = 1
                    else:
                        employed[i, t] = 0
                else:
                    # was employed -> loses job with probability sigma
                    if uniforms[i, t] < sigma:
                        employed[i, t] = 0
                    else:
                        employed[i, t] = 1

    return employed


employed = simulate_employment(S_e_i, uniforms)
employed.shape

# Validation step
print("\nComparing unemployment rate to theoretical:")

t_values = [1, 5, 8, 10, 20, 30, 40, n_years - 1]

theoretical_unemployment = sigma / (sigma + lam)

for t in t_values:
    labor_force_count = 0
    unemployed_count = 0

    for i in range(N):
        if t >= S_e_i[i]:
            # this person has finished school and is in the labor market
            labor_force_count = labor_force_count + 1
            if employed[i, t] == 0:
                unemployed_count = unemployed_count + 1

    unemployment_rate = unemployed_count / labor_force_count
    print(f"t = {t} (age {age_start + t}): unemployment rate = {unemployment_rate:.4f}")

print(f"\nTheoretical steady-state unemployment rate: sigma/(sigma+lam) = {theoretical_unemployment:.4f}")


# Income calculation
income = np.empty((N, n_years))
human_capital = np.empty((N, n_years))

last_job_income = np.empty(N)            # income earned in the most recent job
ever_employed = np.zeros(N, dtype=bool)  # has this person held a job yet?

for i in range(N):
    for t in range(n_years):

        if t < S_e_i[i]:
            # still in education: receives the student grant
            income[i, t] = y_SU
            human_capital[i, t] = 0

        else:
            # in the labor market: first update human capital for this period
            if t == S_e_i[i]:
                # first period after finishing school: starts at the education-specific level
                human_capital[i, t] = h_e0[education_level_i[i]]
            elif employed[i, t] == 1:
                # employed: human capital grows at the education-specific rate, times the shock
                human_capital[i, t] = human_capital[i, t - 1] * (1 + delta_e[education_level_i[i]]) * psi[i, t]
            else:
                # unemployed: human capital depreciates, times the shock
                human_capital[i, t] = human_capital[i, t - 1] * (1 - delta) * psi[i, t]

            # then set income based on employment status
            if employed[i, t] == 1:
                income[i, t] = human_capital[i, t]
                last_job_income[i] = income[i, t]
                ever_employed[i] = True
            else:
                if ever_employed[i]:
                    # unemployed with a previous job: replacement rate of that job's income
                    income[i, t] = rho * last_job_income[i]
                else:
                    # unemployed and never had a job: use the floor
                    income[i, t] = y_floor

income.shape

print("")
print("# 2.2.2")
print("")

print("Plotted mean income and four different deciles over the period")
ages = np.arange(age_start, age_retire)

mean_income = income.mean(axis=0)

p10 = np.percentile(income, 10, axis=0)
p35 = np.percentile(income, 35, axis=0)
p80 = np.percentile(income, 80, axis=0)
p90 = np.percentile(income, 90, axis=0)

plt.figure(figsize=(10, 6))
plt.plot(ages, mean_income, label='Mean', color='black', linewidth=2)
plt.plot(ages, p10, label='10th percentile', linestyle='--')
plt.plot(ages, p35, label='35th percentile', linestyle='--')
plt.plot(ages, p80, label='80th percentile', linestyle='--')
plt.plot(ages, p90, label='90th percentile', linestyle='--')

plt.xlabel('Age')
plt.ylabel('Income')
plt.title('Income over the life cycle: mean and selected quantiles')
plt.legend()
plt.show()

print("# 2.2.3")
print("")
print("Histogram of the income distribution at ages 25, 35, 45, 60")
ages_to_plot = [25, 35, 45, 60]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for idx, age in enumerate(ages_to_plot):
    t = age - age_start  # convert age to the corresponding period index
    axes[idx].hist(income[:, t], bins=50)
    axes[idx].axvline(income[:, t].mean(), color='red', label='Mean')
    axes[idx].set_title(f'Income distribution at age {age}')
    axes[idx].set_xlabel('Income')
    axes[idx].set_ylabel('Count')
    axes[idx].legend()

plt.tight_layout()
plt.show()

print("The plot shows that as time passes, the histograms shrink in size."
      " This likely happens because the three groups each converge against"
      " individual income levels. However, while the histograms shrink in size,"
      " the relative difference between incomes, i.e. the level of inequality,"
      " seems to increase.")

print("")
print("#### 2.3 ####")
print("")

# Gini coefficient: 2 * the area between the 45-degree line (perfect equality)
# and the cumulated income share (the Lorenz curve)
def gini_coefficient(y):
    sorted_y = np.sort(y)
    n = len(sorted_y)

    cumulative_income_share = np.cumsum(sorted_y) / np.sum(sorted_y)  # cumulated % of total income
    population_share = np.arange(1, n + 1) / n                        # cumulated % of population (the 45-degree line)

    gini = 2 * np.sum(population_share - cumulative_income_share) / n

    return gini


# Gini coefficient for the full simulated sample (all individuals, all ages pooled)
gini_full_sample = gini_coefficient(income.flatten())
print(f"Gini coefficient (FULL SAMPLE, all ages pooled): {gini_full_sample:.4f}")

# Test the Gini function on a uniform distribution: the true value is exactly 1/3
uniform_sample = rng.uniform(0, 1, 100_000)
gini_uniform = gini_coefficient(uniform_sample)
print(f"Gini coefficient (UNIFORM TEST): {gini_uniform:.4f} (theoretical: {1/3:.4f})")

# Standard normal CDF, approximated by simulation
def standard_normal_cdf(z, n_draws=1_000_000):
    draws = rng.standard_normal(n_draws)
    return (draws <= z).mean() 
#.mean counts the fraction of draws less than or equal to the mean

# Test the Gini function on a lognormal distribution with s=2:
s = 2
lognormal_sample = rng.lognormal(0, s, size=100_000)
gini_lognormal = gini_coefficient(lognormal_sample)
theoretical_lognormal = 2 * standard_normal_cdf(z = s / np.sqrt(2)) - 1
print(f"Gini coefficient (LOGNORMAL TEST, s={s}): {gini_lognormal:.4f} (theoretical: {theoretical_lognormal:.4f})")


# Lorenz curve for the full sample
sorted_income = np.sort(income.flatten())
n = len(sorted_income)

cumulative_income_share = np.cumsum(sorted_income) / np.sum(sorted_income)
population_share = np.arange(1, n + 1) / n

print("")
print("Plotted Lorenz curve:")

plt.figure(figsize=(8, 8))
plt.plot(population_share, cumulative_income_share, label='Lorenz curve')
plt.plot([0, 1], [0, 1], label='45-degree line (perfect equality)', linestyle='--', color='black')

plt.xlabel('Cumulative share of population')
plt.ylabel('Cumulative share of income')
plt.title('Lorenz curve: full sample')
plt.legend()
plt.show()

# Gini coefficient for every single age
gini_by_age_baseline = np.empty(n_years)

for t in range(n_years):
    gini_by_age_baseline[t] = gini_coefficient(income[:, t])

print("Plotted gini coefficient across all age groups:")

plt.figure(figsize=(10, 6))
plt.plot(ages, gini_by_age_baseline)
plt.xlabel('Age')
plt.ylabel('Gini coefficient')
plt.title('Gini coefficient over the life cycle')
plt.show()

print("The graph shows that inequality increases with age due to differences in human capital "
      "accumulation. Thus, for some age groups the gini-coefficient is larger and for some "
      "smaller than for the entire population.")

print("")
print("#### 2.4 ####")
print("")

print("Scenario: no educational differences.")
# Everyone gets the same (population-weighted average) years of schooling,
# initial human capital, and human capital growth rate, instead of these
# varying by education level.
S_e_avg = round(np.dot(p_e, S_e))
h_e0_avg = np.dot(p_e, h_e0)
delta_e_avg = np.dot(p_e, delta_e)

S_e_i_no_education = np.full(N, S_e_avg)

employed_no_education = simulate_employment(S_e_i_no_education, uniforms)

income_no_education = np.empty((N, n_years))
human_capital_no_education = np.empty((N, n_years))

last_job_income_no_education = np.empty(N)
ever_employed_no_education = np.zeros(N, dtype=bool)

# Income for each year in the "no education" scenario

for i in range(N):
    for t in range(n_years):

        if t < S_e_i_no_education[i]:
            income_no_education[i, t] = y_SU
            human_capital_no_education[i, t] = 0

        else:
            if t == S_e_i_no_education[i]:
                human_capital_no_education[i, t] = h_e0_avg
            elif employed_no_education[i, t] == 1:
                human_capital_no_education[i, t] = human_capital_no_education[i, t - 1] * (1 + delta_e_avg) * psi[i, t]
            else:
                human_capital_no_education[i, t] = human_capital_no_education[i, t - 1] * (1 - delta) * psi[i, t]

            if employed_no_education[i, t] == 1:
                income_no_education[i, t] = human_capital_no_education[i, t]
                last_job_income_no_education[i] = income_no_education[i, t]
                ever_employed_no_education[i] = True
            else:
                if ever_employed_no_education[i]:
                    income_no_education[i, t] = rho * last_job_income_no_education[i]
                else:
                    income_no_education[i, t] = y_floor

gini_full_no_education = gini_coefficient(income_no_education.flatten())
gini_by_age_no_education = np.empty(n_years)
for t in range(n_years):
    gini_by_age_no_education[t] = gini_coefficient(income_no_education[:, t])

print(f"Gini coefficient (no educational differences, full sample): {gini_full_no_education:.4f}")

print("")
print("Scenario: no shocks to human capital.")
# Employment is unaffected by human capital, so the baseline `employed` array is reused.
psi_no_shocks = np.ones((N, n_years))

income_no_shocks = np.empty((N, n_years))
human_capital_no_shocks = np.empty((N, n_years))

last_job_income_no_shocks = np.empty(N)
ever_employed_no_shocks = np.zeros(N, dtype=bool)

for i in range(N):
    for t in range(n_years):

        if t < S_e_i[i]:
            income_no_shocks[i, t] = y_SU
            human_capital_no_shocks[i, t] = 0

        else:
            if t == S_e_i[i]:
                human_capital_no_shocks[i, t] = h_e0[education_level_i[i]]
            elif employed[i, t] == 1:
                human_capital_no_shocks[i, t] = human_capital_no_shocks[i, t - 1] * (1 + delta_e[education_level_i[i]]) * psi_no_shocks[i, t]
            else:
                human_capital_no_shocks[i, t] = human_capital_no_shocks[i, t - 1] * (1 - delta) * psi_no_shocks[i, t]

            if employed[i, t] == 1:
                income_no_shocks[i, t] = human_capital_no_shocks[i, t]
                last_job_income_no_shocks[i] = income_no_shocks[i, t]
                ever_employed_no_shocks[i] = True
            else:
                if ever_employed_no_shocks[i]:
                    income_no_shocks[i, t] = rho * last_job_income_no_shocks[i]
                else:
                    income_no_shocks[i, t] = y_floor

gini_full_no_shocks = gini_coefficient(income_no_shocks.flatten())
gini_by_age_no_shocks = np.empty(n_years)
for t in range(n_years):
    gini_by_age_no_shocks[t] = gini_coefficient(income_no_shocks[:, t])

print(f"Gini coefficient (no human capital shocks, full sample): {gini_full_no_shocks:.4f}")

print("")
print("Scenario: no depreciation of human capital while unemployed.")
delta_no_depreciation = 0.0

income_no_depreciation = np.empty((N, n_years))
human_capital_no_depreciation = np.empty((N, n_years))

last_job_income_no_depreciation = np.empty(N)
ever_employed_no_depreciation = np.zeros(N, dtype=bool)


for i in range(N):
    for t in range(n_years):

        if t < S_e_i[i]:
            income_no_depreciation[i, t] = y_SU
            human_capital_no_depreciation[i, t] = 0

        else:
            if t == S_e_i[i]:
                human_capital_no_depreciation[i, t] = h_e0[education_level_i[i]]
            elif employed[i, t] == 1:
                human_capital_no_depreciation[i, t] = human_capital_no_depreciation[i, t - 1] * (1 + delta_e[education_level_i[i]]) * psi[i, t]
            else:
                human_capital_no_depreciation[i, t] = human_capital_no_depreciation[i, t - 1] * (1 - delta_no_depreciation) * psi[i, t]

            if employed[i, t] == 1:
                income_no_depreciation[i, t] = human_capital_no_depreciation[i, t]
                last_job_income_no_depreciation[i] = income_no_depreciation[i, t]
                ever_employed_no_depreciation[i] = True
            else:
                if ever_employed_no_depreciation[i]:
                    income_no_depreciation[i, t] = rho * last_job_income_no_depreciation[i]
                else:
                    income_no_depreciation[i, t] = y_floor

gini_full_no_depreciation = gini_coefficient(income_no_depreciation.flatten())
gini_by_age_no_depreciation = np.empty(n_years)
for t in range(n_years):
    gini_by_age_no_depreciation[t] = gini_coefficient(income_no_depreciation[:, t])

print(f"Gini coefficient (no depreciation while unemployed, full sample): {gini_full_no_depreciation:.4f}")

print("")
print("Scenario: no unemployment. Everyone is employed in every period once they finish school")

employed_no_unemployment = np.empty((N, n_years))

for i in range(N):
    for t in range(n_years):
        if t < S_e_i[i]:
            employed_no_unemployment[i, t] = 0
        else:
            employed_no_unemployment[i, t] = 1

income_no_unemployment = np.empty((N, n_years))
human_capital_no_unemployment = np.empty((N, n_years))


for i in range(N):
    for t in range(n_years):

        if t < S_e_i[i]:
            income_no_unemployment[i, t] = y_SU
            human_capital_no_unemployment[i, t] = 0

        else:
            if t == S_e_i[i]:
                human_capital_no_unemployment[i, t] = h_e0[education_level_i[i]]
            else:
                human_capital_no_unemployment[i, t] = human_capital_no_unemployment[i, t - 1] * (1 + delta_e[education_level_i[i]]) * psi[i, t]

            income_no_unemployment[i, t] = human_capital_no_unemployment[i, t]

gini_full_no_unemployment = gini_coefficient(income_no_unemployment.flatten())
gini_by_age_no_unemployment = np.empty(n_years)
for t in range(n_years):
    gini_by_age_no_unemployment[t] = gini_coefficient(income_no_unemployment[:, t])

print(f"Gini coefficient (no unemployment, full sample): {gini_full_no_unemployment:.4f}")

print("")
print("Combined plot: age-specific Gini coefficient, baseline vs. all 4 switched-off scenarios")
plt.figure(figsize=(10, 6))
plt.plot(ages, gini_by_age_baseline, label='Baseline', color='black', linewidth=2)
plt.plot(ages, gini_by_age_no_education, label='No educational differences')
plt.plot(ages, gini_by_age_no_shocks, label='No human capital shocks')
plt.plot(ages, gini_by_age_no_depreciation, label='No depreciation while unemployed')
plt.plot(ages, gini_by_age_no_unemployment, label='No unemployment')

plt.xlabel('Age')
plt.ylabel('Gini coefficient')
plt.title('Gini coefficient over the life cycle: baseline vs. switched-off mechanisms')
plt.legend()
plt.show()

print("")
print("In the case of no depreciation of human capital while unemployed, the results"
      " show that the gini-coefficient across the entire population increases, while it is"
      " lower than or equal to the base case within every age group. This is likely caused by"
      " the increased difference between younger and older age groups, since the lack of"
      " depreciation means older workers now simply accumulate more human capital on average.")
print("")

print("#### 2.5 ####")
print("")
print("""To mimic the fact that less educated people more often than highly educated people
do physical work, we augment the model by adding a risk of injury while working.
In this case, human capital depreciates at a faster rate, but the worker is compensated
fully, i.e. receives the same salary as if they had still been working. Note that once the
worker returns to the labour market, they are again compensated relative to their level
of human capital.""")
print("")
print("We set delta_injury = 0.1")
print("We set change of recovery 0.95")
print("Risk of injury is set to [0.05, 0.02, 0.01] with 0.05 = p injury_low")
print("")

p_injury_e = np.array([0.05, 0.02, 0.01])  # probability of injury while employed, by education (short, medium, long)
p_recovery = 0.95                        # probability of recovering from injury each period
delta_injury = 0.1                       # human capital depreciation rate while injured


status = np.empty((N, n_years))
human_capital_injury = np.empty((N, n_years))
income_injury = np.empty((N, n_years))

last_job_income_injury = np.empty(N)
ever_employed_injury = np.zeros(N, dtype=bool)

uniforms_injury = rng.uniform(0.0, 1.0, size=(N, n_years))

for i in range(N):
    for t in range(n_years):

        if t < S_e_i[i]:
            # still in education
            income_injury[i, t] = y_SU
            human_capital_injury[i, t] = 0

        else:
            # Status:
            if t == S_e_i[i]:
                # just finished education: enters labor market as unemployed
                status[i, t] = 0
            else:
                prev_status = status[i, t - 1]
                draw = uniforms_injury[i, t]

                if prev_status == 1:
                    # was employed: could become injured, lose the job, or stay employed
                    p_inj = p_injury_e[education_level_i[i]]
                    if draw < p_inj:
                        status[i, t] = 2  # injured
                    elif draw < p_inj + sigma:
                        status[i, t] = 0  # unemployed
                    else:
                        status[i, t] = 1  # stays employed
                elif prev_status == 2:
                    # was injured: recovers with probability p_recovery
                    status[i, t] = 1 if draw < p_recovery else 2
                else:
                    # was unemployed: finds a job with probability lambda
                    status[i, t] = 1 if draw < lam else 0

            # human capital:
            if t == S_e_i[i]:
                human_capital_injury[i, t] = h_e0[education_level_i[i]]
            elif status[i, t] == 1:
                human_capital_injury[i, t] = human_capital_injury[i, t - 1] * (1 + delta_e[education_level_i[i]]) * psi[i, t]
            elif status[i, t] == 2:
                # injured: human capital depreciates faster than while unemployed
                human_capital_injury[i, t] = human_capital_injury[i, t - 1] * (1 - delta_injury) * psi[i, t]
            else:
                human_capital_injury[i, t] = human_capital_injury[i, t - 1] * (1 - delta) * psi[i, t]

            # income:
            if status[i, t] == 1:
                income_injury[i, t] = human_capital_injury[i, t]
                last_job_income_injury[i] = income_injury[i, t]
                ever_employed_injury[i] = True
            elif status[i, t] == 2:
                # injured: fully compensated at the salary from just before the injury
                income_injury[i, t] = last_job_income_injury[i]
            else:
                if ever_employed_injury[i]:
                    income_injury[i, t] = rho * last_job_income_injury[i]
                else:
                    income_injury[i, t] = y_floor

gini_full_injury = gini_coefficient(income_injury.flatten())
gini_by_age_injury = np.empty(n_years)
for t in range(n_years):
    gini_by_age_injury[t] = gini_coefficient(income_injury[:, t])

print(f"Gini coefficient (injury risk extension, full sample): {gini_full_injury:.4f}")
print("")
print("Thus, we find that this addition increases inequality, however only slighty")


```
